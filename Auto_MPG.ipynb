```python
df=pd.read_csv("Auto MPG.csv")
df.head()
```

<h1 style='color : purple;'>Exploratory Data Analysis</h1>

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x='mpg', y='weight')

plt.title("Relationship between mpg and weight")
plt.xlabel('WEIGHT')
plt.ylabel("MPG")

plt.show()
```

```python
coleration=df['weight'].corr(df['mpg'])
coleration
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x='mpg', y='horsepower')

plt.title("Relationship between mpg and horsepower ")
plt.xlabel('MPG')
plt.ylabel("HORSEPOWER")

plt.show()

df['horsepower'].corr(df['mpg'])
```
